```python
#Importing a Python Libarary
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Loading a dataset
df1 = pd.read_csv('/content/Unemployment in India.csv')

df2 = pd.read_csv('/content/Unemployment_Rate_upto_11_2020.csv')
```

```python
#Display the dataset
print("Dataset 1:")
display(df1.head())

print("Dataset 2:")
display(df2.head())
```

```python
#Checking the shape of the data set
print("Dataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)
```

```python
#checking the columns names of the dataset
print("\nDataset 1 Columns:")
print(df1.columns)

print("\nDataset 2 Columns:")
print(df2.columns)
```

```python
#Checking the Information of the dataset
print("Dataset 1 Information:")
df1.info()

print("\nDataset 2 Information:")
df2.info()
```

# Data Cleaning

```python
#checking a missing values in the dataset
print("Missing Values in Dataset 1:")
print(df1.isnull().sum())

print("\nMissing Values in Dataset 2:")
print(df2.isnull().sum())
```

```python
#Removing Missing Values in the Dataset
df1.dropna(how='all', inplace=True)
df2.dropna(how='all', inplace=True)

print("Dataset 1 Shape after cleaning:", df1.shape)
print("Dataset 2 Shape after cleaning:", df2.shape)
```

```python
#Checking Duplicate Records in the Dataset
print("Duplicate Rows in Dataset 1:", df1.duplicated().sum())

print("Duplicate Rows in Dataset 2:", df2.duplicated().sum())
```

```python
#checking the datatype of the Dataset
print("Dataset 1 Data Types:")
print(df1.dtypes)

print("\nDataset 2 Data Types:")
print(df2.dtypes)
```

```python
#Converting the Date into the Datetime
df1[' Date'] = pd.to_datetime(df1[' Date'].str.strip(), dayfirst=True)

df2[' Date'] = pd.to_datetime(df2[' Date'].str.strip(), dayfirst=True)

print("Dataset 1 Date Type:", df1[' Date'].dtype)
print("Dataset 2 Date Type:", df2[' Date'].dtype)
```

```python
#then if cleaning the all dataset then we will check again missing values in the Dataset
print("Dataset 1:")
print(df1.isnull().sum())

print("\nDataset 2:")
print(df2.isnull().sum())
```

```python
#Checking the statistical data information of the Dataset
print("Dataset 1 Statistical Summary:")
display(df1.describe())

print("Dataset 2 Statistical Summary:")
display(df2.describe())
```

```python
#Checking average unemployment rate in the Dataset
print("Dataset 1 Average Unemployment Rate:",
      df1[' Estimated Unemployment Rate (%)'].mean())

print("\nDataset 2 Average Unemployment Rate:",
      df2[' Estimated Unemployment Rate (%)'].mean())
```

```python
#checking unemployment rate by statewise in the india dataset1
state_unemployment_1 = df1.groupby('Region')[' Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

print(state_unemployment_1)
```

```python
#checking unemployment rate by statewise in the india dataset2
state_unemployment_2 = df2.groupby('Region')[' Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

print(state_unemployment_2)
```

# Data Visualization

```python
#creating a graph of average unemployment rate by state(dataset1)
plt.figure(figsize=(12, 6))

state_unemployment_1.plot(kind='bar')

plt.title('Average Unemployment Rate by State (Dataset 1)')
plt.xlabel('States')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#creating a graph of average unemployment rate by state(dataset2)
plt.figure(figsize=(12, 6))

state_unemployment_2.plot(kind='bar')

plt.title('Average Unemployment Rate by State (Dataset 2)')
plt.xlabel('States')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#Comparision a area wise  of estimated unemployment rate of rural vs urban
area_unemployment = df1.groupby('Area')[' Estimated Unemployment Rate (%)'].mean()

print(area_unemployment)
```

```python
#Comparision of  area wise  of estimated unemployment rate of rural vs urban
plt.figure(figsize=(7, 5))

area_unemployment.plot(kind='bar')

plt.title('Rural vs Urban Unemployment Rate')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#checking a monthly unemployment rate of 2019-2020
monthly_unemployment_1 = df1.groupby(df1[' Date'].dt.to_period('M'))[' Estimated Unemployment Rate (%)'].mean()

print(monthly_unemployment_1)
```

```python
#Creating a  monthly unemployment rate graph of 2019-2020
plt.figure(figsize=(12, 6))

monthly_unemployment_1.plot(kind='line', marker='o')

plt.title('Monthly Unemployment Rate in India (2019-2020)')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Checking a monthly unemployement of 2020
monthly_unemployment_2 = df2.groupby(df2[' Date'].dt.to_period('M'))[' Estimated Unemployment Rate (%)'].mean()

print(monthly_unemployment_2)
```

```python
# Creating a monthly unemployement Graph of 2020
plt.figure(figsize=(12, 6))

monthly_unemployment_2.plot(kind='line', marker='o')

plt.title('Monthly Unemployment Rate in India (January - October 2020)')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#creating a graph to visualize a monthly unemployment rate pattern of 2019-2020
monthly_unemployment_1.plot(kind='bar', figsize=(12, 6))

plt.title('Monthly Unemployment Rate Pattern (2019-2020)')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#creating a graph to visualize a monthly unemployment rate pattern of 2019-2020
monthly_unemployment_2.plot(kind='bar', figsize=(12, 6))

plt.title('Monthly Unemployment Rate Pattern (2020)')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

2. State-wise Differences: Haryana and Tripura recorded relatively higher average unemployment rates, while Meghalaya recorded comparatively lower rates in both datasets.

3. Urban vs Rural Unemployment: The average unemployment rate was approximately 13.17% in urban areas and 10.32% in rural areas in Dataset 1.

4. Recovery Pattern: Unemployment rates started declining after May 2020, showing signs of improvement in employment conditions.
